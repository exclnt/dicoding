#### Kalkulasi Interquartile Range (IQR) untuk Identifikasi Outlier

**Deskripsi Kode:**

* **`Q1 = df[numeric_features].quantile(0.25)`**: Menghitung Kuartil 1 (Kuartil Bawah). Proses ini mengambil nilai persentil ke-25 dari seluruh kolom berjenis numerik dalam suatu *dataframe*.
* **`Q3 = df[numeric_features].quantile(0.75)`**: Menghitung Kuartil 3 (Kuartil Atas). Proses ini mengambil nilai persentil ke-75 dari seluruh kolom berjenis numerik dalam suatu *dataframe*.
* **`IQR = Q3 - Q1`**: Menghitung nilai *Interquartile Range* (Rentang Antarkuartil). Proses ini mengukur rentang penyebaran 50% data yang berada di tengah dengan mencari selisih antara Q3 dan Q1. Nilai IQR ini merupakan komponen utama untuk menghitung batas atas dan batas bawah dalam pendeteksian *outlier*.

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1
```

#### Penyaringan Outlier dan Penggabungan Data

**Deskripsi Kode:**

* **`condition = ~((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)`**: Membuat kondisi logika (mask) untuk mendeteksi *outlier*.
* `(Q1 - 1.5 * IQR)` menetapkan batas bawah normal.
* `(Q3 + 1.5 * IQR)` menetapkan batas atas normal.
* Fungsi `.any(axis=1)` mengecek apakah ada setidaknya satu nilai *outlier* dalam satu baris.
* Operator `~` (negasi) membalikkan kondisi, sehingga hanya baris yang **bersih dari *outlier*** yang akan bernilai `True`.


* **`df_filtered_numeric = df.loc[condition, numeric_features]`**: Mengambil baris data yang bernilai `True` (tanpa *outlier*) khusus untuk kolom-kolom bertipe numerik dan menyimpannya dalam variabel baru.
* **`categorical_features = df.select_dtypes(include=['object']).columns`**: Mengekstrak daftar nama kolom yang bertipe data `object` (biasanya berisi data teks atau kategorikal) dari *dataframe*.
* **`df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)`**: Menggabungkan kembali data numerik yang sudah difilter dengan data kategorikal yang difilter menggunakan kondisi baris yang sama. Parameter `axis=1` memastikan penggabungan dilakukan secara horizontal (bersebelahan sebagai kolom baru).

```python
# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]
 
# Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['object']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

**Deskripsi Kode:**

* **`from sklearn.model_selection import train_test_split`**: Mengimpor fungsi `train_test_split` dari modul `model_selection` milik *library* **scikit-learn**. Fungsi ini adalah standar industri untuk membagi dataset secara acak.
* **`x_train, x_test, y_train, y_test = train_test_split(X, y, ...)`**: Memecah kumpulan fitur (`X`) dan target (`y`) menjadi empat variabel terpisah:
* `x_train` dan `y_train`: Kumpulan data yang akan digunakan untuk **melatih** atau mengajarkan model *Machine Learning*.
* `x_test` dan `y_test`: Kumpulan data yang dipisahkan atau disembunyikan dari model selama masa pelatihan, guna digunakan nanti untuk **menguji** seberapa akurat tebakan model tersebut pada data baru.


* **`test_size=0.2`**: Menetapkan proporsi pembagian data. Angka `0.2` berarti **20%** dari keseluruhan baris datasetmu akan dialokasikan sebagai data pengujian (*testing*), sementara mayoritas sisanya, yaitu **80%**, akan digunakan sebagai data pelatihan (*training*).
* **`random_state=1`**: Menetapkan angka acuan (sering disebut *seed*) untuk proses pengacakan saat pembagian data. Dengan menetapkan angka `1`, kamu memastikan bahwa sistem akan melakukan pengacakan baris dengan urutan yang sama persis setiap kali kode ini dijalankan ulang. Hal ini sangat penting agar hasil evaluasi modelmu tetap konsisten dan tidak berubah-ubah saat kamu memodifikasi hal lain di masa depan.

### Modelilng

#### Pelatihan Tiga Model Machine Learning Berbeda

**Deskripsi Kode:**

* **`from sklearn import linear_model`** dan **`lars = linear_model.Lars(n_nonzero_coefs=1).fit(x_train, y_train)`**: Kode ini mengimpor modul model linear dan langsung melatih model pertama menggunakan algoritma **Least Angle Regression (LARS)**.
* Parameter `n_nonzero_coefs=1` menginstruksikan algoritma ini untuk sangat membatasi dirinya dengan hanya menggunakan satu variabel (fitur prediktor) yang paling kuat pengaruhnya terhadap target.
* Fungsi `.fit(x_train, y_train)` adalah perintah wajib bagi mesin untuk mulai "belajar" mencocokkan pola dari data latihmu.


* **`from sklearn.linear_model import LinearRegression`** dan **`LR = LinearRegression().fit(x_train, y_train)`**: Mengimpor dan melatih model kedua menggunakan algoritma **Linear Regression** standar. Ini adalah algoritma regresi klasik yang akan mencoba menarik sebuah garis lurus khayalan terbaik (optimal) yang bisa mewakili tren hubungan antara semua data fitur (`x_train`) dengan target harganya (`y_train`).
* **`from sklearn.ensemble import GradientBoostingRegressor`** beserta **`GBR = GradientBoostingRegressor(random_state=184)`** lalu **`GBR.fit(x_train, y_train)`**: Mengimpor, mempersiapkan, dan melatih model ketiga menggunakan **Gradient Boosting Regressor**.
* Ini termasuk algoritma *ensemble* tingkat lanjut yang sangat tangguh. Cara kerjanya adalah dengan membangun banyak model sederhana (*decision trees*) secara bertahap, di mana setiap model baru yang dibuat akan berfokus untuk mengoreksi kesalahan (eror) tebakan dari model-model sebelumnya.
* Sama seperti pada tahap pembagian data, `random_state=184` digunakan untuk mengunci elemen acak pada algoritma ini agar hasil prediksi akhir modelmu selalu stabil dan bisa direplikasi.

```python
# Melatih model 1 dengan algoritma Least Angle Regression
from sklearn import linear_model
lars = linear_model.Lars(n_nonzero_coefs=1).fit(x_train, y_train)
 
# Melatih model 2 dengan algoritma Linear Regression
from sklearn.linear_model import LinearRegression
LR = LinearRegression().fit(x_train, y_train)
 
# Melatih model 3 dengan algoritma Gradient Boosting Regressor
from sklearn.ensemble import GradientBoostingRegressor
GBR = GradientBoostingRegressor(random_state=184)
GBR.fit(x_train, y_train)
```

#### Evaluasi Model
#### Evaluasi Performa Model dan Pembuatan Tabel Hasil (LARS)

**Deskripsi Kode:**

* **`from sklearn.metrics import ...`**: Mengimpor tiga alat ukur standar evaluasi regresi dari *library* **scikit-learn** (yaitu MAE, MSE, dan R2 Score) untuk menilai seberapa akurat tebakan modelmu.
* **`pred_lars = lars.predict(x_test)`**: Memerintahkan model LARS yang sebelumnya sudah dilatih untuk mulai melakukan ujian sungguhan. Model diminta untuk menebak (memprediksi) nilai target menggunakan kumpulan data ujian (`x_test`). Hasil tebakannya kemudian disimpan di variabel `pred_lars`.
* **Perhitungan Skor Evaluasi:**
* **`mae_lars = mean_absolute_error(y_test, pred_lars)`**: Menghitung **MAE** (*Mean Absolute Error*). Ini adalah rata-rata dari total selisih/kesalahan tebakan. Semakin kecil angkanya, berarti tebakan model semakin akurat.
* **`mse_lars = mean_squared_error(y_test, pred_lars)`**: Menghitung **MSE** (*Mean Squared Error*). Mirip dengan MAE, namun selisih kesalahannya dikuadratkan terlebih dahulu. Ini berguna untuk memberikan "penalti" atau angka yang sangat bengkak jika model membuat tebakan yang melesetnya terlalu jauh.
* **`r2_lars = r2_score(y_test, pred_lars)`**: Menghitung **R-Squared** ($R^2$). Ini adalah skor dari 0 hingga 1 yang mengukur seberapa baik variabel fiturmu bisa menjelaskan variasi harga target. Semakin nilainya mendekati angka 1 (atau 100%), semakin sempurna modelmu.


* **`data = {...}`**: Memasukkan ketiga skor hasil perhitungan (MAE, MSE, R2) tersebut ke dalam sebuah struktur data dasar Python bernama *dictionary* secara rapi.
* **`df_results = pd.DataFrame(data, index=['Lars'])`**: Mengubah data dari *dictionary* tadi ke dalam format tabel (*dataframe*) Pandas agar jauh lebih mudah dibaca secara visual. Parameter `index=['Lars']` memberikan nama baris tersebut sebagai 'Lars' agar kamu tahu bahwa itu adalah hasil evaluasi khusus untuk model LARS.
* **`df_results`**: Memanggil dan menampilkan tabel hasil evaluasi akhir tersebut ke layar konsol.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
 
# Evaluasi pada model LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)
 
# Membuat dictionary untuk menyimpan hasil evaluasi
data = {
    'MAE': [mae_lars],
    'MSE': [mse_lars],
    'R2': [r2_lars]
}
 
# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars'])
df_results
```

#### Menyimpan Model Machine Learning dengan Pickle

**Deskripsi Kode:**

* **`import pickle`**: Mengimpor *library* bawaan standar Python bernama **pickle**. Fungsinya mirip dengan `joblib` sebelumnya, yaitu untuk melakukan *serialization* (mengubah objek kode Python menjadi format data biner/byte yang bisa disimpan ke dalam penyimpanan komputermu).
* **`with open('gbr_model.pkl', 'wb') as file:`**: Blok kode untuk membuka (atau membuat jika belum ada) sebuah *file* baru bernama `'gbr_model.pkl'`.
* Parameter `'wb'` adalah singkatan dari *write-binary*, yang menginstruksikan sistem bahwa kamu akan menulis atau memasukkan data ke dalam *file* tersebut dalam format biner (bukan teks yang bisa dibaca manusia).
* Penggunaan metode `with open(...)` (*context manager*) adalah praktik terbaik di Python karena ia memastikan *file* tersebut akan otomatis ditutup dengan aman segera setelah proses penyimpanan selesai, sehingga menghindari risiko *file* *corrupt* atau kebocoran memori komputasi.


* **`pickle.dump(GBR, file)`**: Mengeksekusi proses penyimpanan. Fungsi ini mengambil objek model Gradient Boosting Regressor milikmu (variabel `GBR`) dan menyuntikkan seluruh strukturnya secara permanen ke dalam `file` yang sedang dibuka tadi.

*Catatan Tambahan:* Baik `joblib` (di kodemu sebelumnya) maupun `pickle` memiliki fungsi dan tujuan akhir yang sama. Perbedaannya, `joblib` biasanya lebih disukai oleh praktisi data karena lebih cepat dan ringan saat menangani model *Machine Learning* besar yang banyak mengandung *array* internal (seperti *array* NumPy), sedangkan `pickle` lebih umum untuk menyimpan objek Python standar apa saja.

```python
import pickle
 
# Menyimpan model ke dalam file
with open('./model/gbr_model.pkl', 'wb') as file:
    pickle.dump(GBR, file)
```

### Deployment dan Monitoring
#### Memuat Kembali (Loading) Model Machine Learning

**Deskripsi Kode:**

* **`joblib_model = joblib.load('./model/gbr_model.joblib')`**: Perintah ini digunakan untuk membaca dan membangkitkan kembali (memuat) model yang sebelumnya sudah kamu simpan menggunakan *library* **joblib**. Model yang ada di dalam file fisik `'gbr_model.joblib'` ditarik masuk ke memori dan disimpan ke dalam variabel baru bernama `joblib_model`. Mulai dari titik ini, variabel tersebut bisa langsung kamu gunakan untuk melakukan prediksi pada data baru.
* **`with open('./model/gbr_model.pkl', 'rb') as file:`**: Membuka file `'gbr_model.pkl'` yang sebelumnya kamu simpan menggunakan metode **pickle**. Perbedaan utamanya ada pada parameter `'rb'` (*read-binary*). Ini menginstruksikan sistem bahwa kamu hanya ingin **membaca** isi dari file biner tersebut, yang merupakan kebalikan dari `'wb'` (*write-binary*) saat proses penyimpanan tadi.
* **`pickle_model = pickle.load(file)`**: Mengeksekusi proses pembacaan data dari dalam file yang sedang dibuka. Fungsi `.load()` akan mengekstrak seluruh wujud model *Machine Learning* tersebut dan menghidupkannya kembali ke dalam variabel `pickle_model`.

```python
# Memuat model dari file joblib
joblib_model = joblib.load('./model/gbr_model.joblib')
 
# Memuat model dari file pickle
with open('./model/gbr_model.pkl', 'rb') as file:
    pickle_model = pickle.load(file)
```
