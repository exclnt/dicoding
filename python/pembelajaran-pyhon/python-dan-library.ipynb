# Python dan Library
## External Libraries
### Library numpy 
NumPy (Numerical Python) digunakan untuk melakukan operasi numerik/matematis, terutama terhadap kumpulan angka.

```python
import numpy as np

angka = np.array([10, 20, 30, 40, 50])

print(angka)
print(angka.mean())
print(angka.sum())
```

mengunakan numpy untuk menghitung determinan

```python
B = np.array([[4, 2],
              [3, 1]])

det_B = np.linalg.det(B)
print(det_B)  # Output: -2.0
```

numpy perjumlahan matrik

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [10, 11, 12]])
C = A + B
print(C)
# Output:
# [[ 8 10 12]
#  [14 16 18]]
```

### Library Pandas
Pandas digunakan untuk mengolah dan menganalisis data berbentuk tabel.

```python
import pandas as pd

data = {'Nama': ['Alice', 'Bob', 'Charlie'], 'Usia': [25, 30, 35]}
df = pd.DataFrame(data)

print(df)
print(df['Nama'])
```

### LIbrary Mathploid
Matplotlib adalah library Python untuk membuat visualisasi data, seperti:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.plot(x, y)
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')
plt.title('Plot Garis Sederhana')
plt.show()
```
